```python
import pickle
import os
import matplotlib.pyplot as plt

files = [file for file in os.listdir() if file.endswith(".pkl")]
mus = []
sigs = []
for file in files:
    motor_data = pickle.load(open(file, 'rb'))
    #print(motor_data.keys())
    if int(motor_data["motor_calibration_power"]) == 45:
        mus.append(motor_data['motor_mu_in'])
        sigs.append(motor_data['motor_sigma_in'])

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].hist(mus, bins=10)
axs[0].set_title("Distribution of the motor average dps during calibration")

axs[1].hist(sigs, bins=10)
axs[1].set_title("Distribution of the motor average std during calibration")

plt.show()
```
